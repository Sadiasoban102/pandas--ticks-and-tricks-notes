# 16-Spliting string into multiple columns

```python
#import libraries
import pandas as pd

df= pd.DataFrame({'name': ['sadia Ghazanfar', 'Minhal Sherazi'],
                  'location': ['Lodhran, Pakistan', 'Sheikupura, Pakistan'] })
df
```

```python
# splitting into 2 columns
df.name.str.split(' ',expand=True)
```

```python
#adding those splits into new columns
df[["first_name", "last_name"]]= df.name.str.split(' ',expand=True)
df
```

```python
df.location.str.split(' ', expand=True)
```

```python
df[["city", "country"]]= df.location.str.split(',',expand=True)
df
```

```python
# refine datasets
df = df[['first_name','last_name','city','country']]
df
```
